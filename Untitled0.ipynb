```python
n = int(input("Enter an integer: "))

n = abs(n)
sum_digits = 0

while n > 0:
    sum_digits += n % 10
    n //= 10

print("Sum of digits:", sum_digits)
```

```python

num = int(input("Enter an integer: "))
num = abs(num)
if num == 0:
    product_digits = 0
else:
    product_digits = 1

temp_num = num

while temp_num > 0:
    digit = temp_num % 10
    product_digits *= digit
    temp_num //= 10

print("Product of digits:", product_digits)
```

```python
num = int(input("Enter an integer to reverse: "))
sign = 1
if num < 0:
    sign = -1
    num = abs(num)

reversed_num = 0
original_num = num

while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num //= 10
reversed_num *= sign

print(f"The original number was: {original_num}")
print(f"The reversed number is: {reversed_num}")
```

```python
num = int(input("Enter an integer to check for palindrome: "))
original_num = num
abs_num = abs(num)

reversed_num = 0
temp_num = abs_num

while temp_num > 0:
    digit = temp_num % 10
    reversed_num = reversed_num * 10 + digit
    temp_num //= 10

if abs_num == reversed_num:
    print(f"{original_num} is a palindrome number.")
else:
    print(f"{original_num} is not a palindrome number.")
```

```python
num = int(input("Enter an integer to check for Armstrong number: "))

original_num = num
abs_num = abs(num)

num_digits = 0
temp_num = abs_num
if temp_num == 0:
    num_digits = 1
else:
    while temp_num > 0:
        temp_num //= 10
        num_digits += 1

sum_of_powers = 0
temp_num = abs_num

while temp_num > 0:
    digit = temp_num % 10
    sum_of_powers += digit ** num_digits
    temp_num //= 10

if abs_num == sum_of_powers:
    print(f"{original_num} is an Armstrong number.")
else:
    print(f"{original_num} is not an Armstrong number.")
```
